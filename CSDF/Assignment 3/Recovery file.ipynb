```python
import os
import shutil

def recover_deleted_files(source, destination):
    """
    Simple File Recovery Script
    Recovers deleted files from a Recycle Bin/Trash folder to a new folder.
    """
    # Create destination folder if not exists
    if not os.path.exists(destination):
        os.makedirs(destination)

    print(f"[+] Recovering files from {source} to {destination}...")

    recovered_count = 0
    for root, dirs, files in os.walk(source):
        for file in files:
            file_path = os.path.join(root, file)
            dest_path = os.path.join(destination, file)
            try:
                shutil.copy2(file_path, dest_path)
                recovered_count += 1
                print(f"[Recovered] {file}")
            except Exception as e:
                print(f"[Error] Could not recover {file}: {e}")

    print(f"\n✅ Recovery complete! Total files recovered: {recovered_count}")

# Example usage (Windows Recycle Bin path)
if __name__ == "__main__":
    # NOTE: Replace this with your actual path if different
    recycle_bin_path = r"C:\$Recycle.Bin"
    recovery_output_path = r"C:\Users\Akhila\Documents\RecoveredFiles"

    recover_deleted_files(recycle_bin_path, recovery_output_path)
```
